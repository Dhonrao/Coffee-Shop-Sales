Q1. What are the total sales made by the coffee shop?

```python
df['Total_Sales'] = df['transaction_qty'] * df['unit_price']
```

```python
total_sales = df['Total_Sales'].sum()
```

```python
total_sales
```

Q2. Which product categories generate the most revenue?

```python
df['product_category'].unique()
```

```python
df['product_type'].unique()
```

```python
df['product_detail'].unique()
```

```python
product_sales = df.groupby('product_category')['Total_Sales'].sum()
```

```python
product_sales.sort_values(ascending A= False).head(10)
```

```python
plt.figure(figsize=(10,5))
product_sales.plot(kind = 'bar',color = 'blue')
plt.title('Top 10 Selling Products')
plt.show()
```

Q3. Which store/location performs the best in terms of sales?

```python
store_sales = df.groupby('store_location')['Total_Sales'].sum().sort_values(ascending = False)
```

```python
store_sales
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=store_sales.index, y=store_sales.values, palette='Set1')
plt.title('Total Sales by Store')
plt.show()
```

```python
plt.figure(figsize=(7,7))
plt.pie(store_sales,labels = store_sales.index,autopct = '%1.2f%%',startangle = 140, shadow = True)
plt.title("Sales Contribution by Store Location")
plt.show()
```

Q3. What day of the week has the highest sales?

Q5. Which products are most popular

```python
popular_product
```

Q6 Are there any peak sales hours during the day?

```python
hourly_sales = df.groupby('Hour')['Total_Sales'].sum().sort_index()
```

```python
hourly_sales
```

```python
plt.figure(figsize=(10,5))
plt.plot(hourly_sales.index, hourly_sales.values, marker='o', color='teal')
plt.title("Sales by Hour of the Day")
plt.xlabel("Hour of the Day (0–23)")
plt.ylabel("Total Sales")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.bar(hourly_sales.index, hourly_sales.values, color='coral')
plt.title("Total Sales by Hour of the Day")
plt.xlabel("Hour of the Day (0–23)")
plt.ylabel("Total Sales")
plt.show()
```

Q7 Which Store Location Generates the Highest Sales?

```python
store_sales = df.groupby('store_location')['Total_Sales'].sum().sort_values(ascending=False)
```

```python
store_sales
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=store_sales.index, y=store_sales.values, palette='viridis')
plt.title("Total Sales by Store Location")
plt.xlabel("Store Location")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

Q11: “Which product type is most popular (sold most frequently)?”
